This notebook reads xls national and subnational population tables published by the UK Office of National Statistics (ONS) and produces an xarray Dataset to be read by the AgriFoodPy package when importing the population modules.

The data processed here represents mid-year estimates obtained for the year 2024.

``` python
from agrifoodpy.population.population import UK_ONS_MYE24UK
```

The dimension coordinates of the array are
- **Code** : GSS code administrative unit identifyer

Additional label coordinates aligned with the above dimension coordinate
- **Name** : Name of the administrative unit

The original xls files processed in this notebook can be found here:
https://www.ons.gov.uk/peoplepopulationandcommunity/populationandmigration/populationestimates/datasets/populationestimatesforukenglandandwalesscotlandandnorthernireland



````python
import pandas as pd
import xarray as xr
import numpy as np
````

````python
datafile = "mye24tablesuk.xlsx"
````

````python
df = pd.read_excel(datafile, sheet_name="MYE2 - Persons", skiprows=7, dtype={"Code": str, "Name": str})
df.head()
````

````python
ds = xr.Dataset.from_dataframe(df.set_index(['Code']))["All ages"]
ds = ds.assign_coords(Name=("Code", df.set_index("Code")["Name"].values))

ds["Code"] = ds["Code"].astype(str)
ds["Name"] = ds["Name"].astype(str)

ds
````

````python
ds.to_netcdf("../data/UK_ONS_MYE24UK.nc")
````
